```python
BRONZE_TABLE = "medical_catalog.bronze.medical_transcriptions"
df_bronze = spark.table(BRONZE_TABLE)
print("Total rows:", df_bronze.count())
```

```python
from pyspark.sql.functions import col, count, when, isnull

null_counts = df_bronze.select([
    count(when(isnull(c) | (col(c) == ""), c)).alias(c)
    for c in ["no", "description", "medical_specialty", "sample_name", "transcription", "keywords"]
])
display(null_counts)
```

```python
total = df_bronze.count()
distinct = df_bronze.dropDuplicates(["no"]).count()
print("Total rows:", total)
print("Distinct rows:", distinct)
print("Duplicate rows:", total - distinct)
```

```python
display(
    df_bronze.groupBy("medical_specialty")
             .count()
             .orderBy("count", ascending=False)
)
```

```python
from pyspark.sql.functions import length

display(
    df_bronze.select(
        length("transcription").alias("transcription_length")
    ).summary()
)
```
